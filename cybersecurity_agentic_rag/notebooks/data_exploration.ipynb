# 1. Data Exploration

This notebook is for exploring the synthetic cybersecurity dataset. We will load the data, examine its properties, and visualize the distribution of different event types.

## 1.1. Load Libraries and Data

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set plot style
sns.set_style("whitegrid")

# Load the dataset
data_path = '../data/processed/synthetic_cyber_events.csv'
try:
    df = pd.read_csv(data_path)
    print("Dataset loaded successfully!")
except FileNotFoundError:
    print(f"Error: Dataset not found at {data_path}. Please run the data generation script first.")
```

## 1.2. Initial Data Inspection

```python
if 'df' in locals():
    # Display the first few rows
    print("First 5 rows of the dataset:")
    display(df.head())
    
    # Display basic information about the dataframe
    print("\nDataset Info:")
    df.info()
    
    # Display summary statistics
    print("\nSummary Statistics:")
    display(df.describe(include='all'))
```

## 1.3. Visualize Data Distributions

```python
if 'df' in locals():
    # Plot the distribution of event types
    plt.figure(figsize=(10, 6))
    sns.countplot(y=df['event_type'], order=df['event_type'].value_counts().index)
    plt.title('Distribution of Event Types', fontsize=16)
    plt.xlabel('Count', fontsize=12)
    plt.ylabel('Event Type', fontsize=12)
    plt.tight_layout()
    plt.show()
    
    # Plot the distribution of malicious vs. benign events
    plt.figure(figsize=(7, 5))
    sns.countplot(x=df['is_malicious_ground_truth'])
    plt.title('Distribution of Malicious vs. Benign Events', fontsize=16)
    plt.xlabel('Is Malicious (Ground Truth)', fontsize=12)
    plt.ylabel('Count', fontsize=12)
    plt.xticks([0, 1], ['Benign', 'Malicious'])
    plt.tight_layout()
    plt.show()
```
