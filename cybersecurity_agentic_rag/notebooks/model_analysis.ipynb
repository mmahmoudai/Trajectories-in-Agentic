## 2.1. Analyzing the RL Agent's Performance

Here, we would load the trained RL agent and inspect its behavior. This could involve:
- Plotting the learning curve (rewards over time) from training logs.
- Testing the agent's decisions on a set of hand-crafted scenarios.
- Visualizing the agent's policy if possible.

```python
import os
from stable_baselines3 import PPO
from cybersecurity_agentic_rag.src.rl.environment import CybersecurityEnv

model_path = '../models/ppo_cyber_agent.zip'

if os.path.exists(model_path):
    print(f"Loading trained RL model from {model_path}")
    rl_model = PPO.load(model_path)
    print("RL model loaded successfully.")
    
    # Example: Test the model on a sample observation
    env = CybersecurityEnv()
    obs, _ = env.reset()
    action, _ = rl_model.predict(obs, deterministic=True)
    print(f"\nFor observation: {obs}\nPredicted action: {action} (0:Ignore, 1:Quarantine, 2:Block)")
else:
    print(f"RL model not found at {model_path}. Please run the training script first.")
```
