# 3. Results Visualization

This notebook is for generating the final plots and tables for the research paper based on the results from the experiment scripts.

## 3.1. Load Libraries and Results Data

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Set plot style
sns.set_theme(style="whitegrid", palette="viridis")

# Create directory for final figures
output_dir = '../visualizations/'
os.makedirs(output_dir, exist_ok=True)

# Load results from experiments
baseline_results_path = '../experiments/baseline_comparison_results.csv'
ablation_results_path = '../experiments/ablation_study_results.csv'

try:
    baseline_df = pd.read_csv(baseline_results_path)
    ablation_df = pd.read_csv(ablation_results_path)
    print("Experiment results loaded successfully.")
except FileNotFoundError as e:
    print(f"Error: {e}. Please run the experiment scripts first.")
```

## 3.2. Visualize Baseline Comparison

This plot compares the performance of our full agentic RAG system against the baseline RAG system.

```python
if 'baseline_df' in locals():
    plt.figure(figsize=(10, 6))
    
    # Melt the dataframe to have metrics in one column for easier plotting
    baseline_melted = baseline_df.melt(id_vars='System', var_name='Metric', value_name='Score')
    
    ax = sns.barplot(data=baseline_melted, x='Metric', y='Score', hue='System')
    
    plt.title('Performance Comparison: Agentic RAG vs. Baseline', fontsize=16)
    plt.ylabel('Score', fontsize=12)
    plt.xlabel('Metric', fontsize=12)
    plt.ylim(0, 1.05)
    plt.legend(title='System', fontsize=10)
    
    # Add labels to the bars
    for p in ax.patches:
        ax.annotate(f'{p.get_height():.2f}', (p.get_x() + p.get_width() / 2., p.get_height()),
                    ha='center', va='center', xytext=(0, 9), textcoords='offset points')
    
    # Save the figure
    fig_path = os.path.join(output_dir, 'figure_baseline_comparison.png')
    plt.savefig(fig_path, dpi=300, bbox_inches='tight')
    print(f"Figure saved to {fig_path}")
    
    plt.show()
```

## 3.3. Visualize Ablation Study Results

This plot shows the impact on F1-score as we remove components from the full system.

```python
if 'ablation_df' in locals():
    plt.figure(figsize=(12, 7))
    
    # Sort by F1-score for better visualization
    ablation_df_sorted = ablation_df.sort_values('f1_score', ascending=False)
    
    ax = sns.barplot(data=ablation_df_sorted, x='f1_score', y='Configuration')
    
    plt.title('Ablation Study: Impact of Components on F1-Score', fontsize=16)
    plt.xlabel('F1-Score', fontsize=12)
    plt.ylabel('System Configuration', fontsize=12)
    plt.xlim(0, 1.0)
    
    # Add labels to the bars
    for p in ax.patches:
        width = p.get_width()
        plt.text(width, p.get_y() + p.get_height() / 2, f'{width:.3f}', va='center')

    # Save the figure
    fig_path = os.path.join(output_dir, 'figure_ablation_study.png')
    plt.savefig(fig_path, dpi=300, bbox_inches='tight')
    print(f"Figure saved to {fig_path}")
    
    plt.show()
```
